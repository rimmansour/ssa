```python
import csv
import json
def export_flat_csv(jsonl_path, csv_path):
    cols = [
        "agence_id", "adresse_input", "status", "place_name", "place_address",
        "place_rating", "place_total_reviews", "place_url", "lat", "lng",
        "review_id", "author", "author_info", "rating", "date", "text",
        "owner_response", "owner_response_date", "scraped_at",
    ]
    with open(jsonl_path, encoding="utf-8") as fin, \
         open(csv_path, "w", newline="", encoding="utf-8-sig") as fout:
        w = csv.DictWriter(fout, fieldnames=cols, delimiter=";")
        w.writeheader()
        for line in fin:
            a = json.loads(line)
            p = a.get("place", {})
            base = {
                "agence_id": a["id"], "adresse_input": a["adresse_input"],
                "status": a["status"], "place_name": p.get("name"),
                "place_address": p.get("address"), "place_rating": p.get("rating"),
                "place_total_reviews": p.get("total_reviews"), "place_url": p.get("url"),
                "lat": p.get("lat"), "lng": p.get("lng"),
                "scraped_at": a["scraped_at"],
            }
            if not a["reviews"]:
                w.writerow(base)
            for r in a["reviews"]:
                w.writerow({**base, **{k: r.get(k) for k in (
                    "review_id", "author", "author_info", "rating", "date",
                    "text", "owner_response", "owner_response_date")}})
OUTPUT_JSONL = "avis_agences1.jsonl"
OUTPUT_CSV = "avis_agences1.csv"
export_flat_csv(OUTPUT_JSONL, OUTPUT_CSV)
import pandas as pd
df = pd.read_csv('avis_agences1.csv', sep=';', encoding='utf-8-sig'  )
df[df.status=='ok'].to_csv('avis_agences1_ok.csv', index=False, encoding='utf-8-sig',sep=';')
```

```python
import csv
import json

REVIEW_KEYS = (
    "review_id", "author", "author_info", "rating", "date",
    "text", "owner_response", "owner_response_date",
)
PLACE_KEYS = {  # clé JSONL -> colonne CSV
    "name": "place_name",
    "address": "place_address",
    "rating": "place_rating",
    "total_reviews": "place_total_reviews",
    "url": "place_url",
    "lat": "lat",
    "lng": "lng",
}


def _val(v):
    """Chaîne vide -> None (le CSV a écrit None comme une cellule vide)."""
    return None if v is None or v == "" else v


def _to_int(v):
    v = _val(v)
    if v is None:
        return None
    try:
        return int(float(v))
    except ValueError:
        return v


def _to_float(v):
    v = _val(v)
    if v is None:
        return None
    try:
        return float(v.replace(",", "."))
    except ValueError:
        return v


def _to_id(v):
    """Garde l'id en int s'il ressemble à un entier, sinon en str."""
    v = _val(v)
    if v is not None and v.lstrip("-").isdigit():
        return int(v)
    return v


def csv_to_jsonl(csv_path, jsonl_path):
    agences = {}  # dict ordonné : conserve l'ordre d'apparition

    with open(csv_path, encoding="utf-8-sig", newline="") as fin:
        for row in csv.DictReader(fin, delimiter=";"):
            aid = row["agence_id"]

            if aid not in agences:
                place = {
                    "name": _val(row["place_name"]),
                    "address": _val(row["place_address"]),
                    "rating": _val(row["place_rating"]),
                    "total_reviews": _to_int(row["place_total_reviews"]),
                    "url": _val(row["place_url"]),
                    "lat": _to_float(row["lat"]),
                    "lng": _to_float(row["lng"]),
                }
                # Si aucune info de lieu, on retombe sur {} comme a.get("place", {})
                if all(v is None for v in place.values()):
                    place = {}

                agences[aid] = {
                    "id": _to_id(aid),
                    "adresse_input": row["adresse_input"],
                    "status": row["status"],
                    "place": place,
                    "reviews": [],
                    "scraped_at": row["scraped_at"],
                }

            # Une ligne sans aucun champ d'avis = agence sans reviews
            if any(_val(row[k]) is not None for k in REVIEW_KEYS):
                agences[aid]["reviews"].append({
                    "review_id": _val(row["review_id"]),
                    "author": _val(row["author"]),
                    "author_info": _val(row["author_info"]),
                    "rating": _to_int(row["rating"]),
                    "date": _val(row["date"]),
                    "text": _val(row["text"]),
                    "owner_response": _val(row["owner_response"]),
                    "owner_response_date": _val(row["owner_response_date"]),
                })

    with open(jsonl_path, "w", encoding="utf-8") as fout:
        for a in agences.values():
            fout.write(json.dumps(a, ensure_ascii=False) + "\n")


INPUT_CSV = "avis_agences3_ok.csv"
OUTPUT_JSONL = "avis_agences3_ok.jsonl"
csv_to_jsonl(INPUT_CSV, OUTPUT_JSONL)
```
